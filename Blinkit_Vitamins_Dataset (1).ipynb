```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import glob
import plotly.express as px
```

```python
df = pd.read_csv('blinkit_products.csv')
print(df)
```

```python
# data cleaning
df = df.drop_duplicates()
df.columns= df.columns.str.strip()
df = df.dropna()
print(df)
```

```python
print(df.head())
```

```python
# Category Distribution
category = df["category"].value_counts()

px.pie(
    values=category.values,
    names=category.index,
    title="Blinkit Category Distribution",
    hole=0.4,
    color_discrete_sequence=px.colors.qualitative.Bold
).show()
```

```python
# select vitamin
vitamins = df[df['product_name'].astype(str).str.contains('vitamin', case=False,na=False)]
display(vitamins)
```

```python
print('Total Records:', len(vitamins))
```

```python
print('Average Price:', vitamins['price'].mean())
```

```python
print('Average MRP:', vitamins ['mrp'].mean())
```

```python
print('Average Margin:', vitamins['margin_percentage'].mean())
```

```python
px.bar(vitamins, x='brand', y='price', color = 'brand', title='Vitamins Price by Brand').show()
```

```python
#Price vs MRP
px.bar(
    vitamins,
    x="product_name",
    y=["price", "mrp"],
    barmode="group",
    title="Vitamins Price vs MRP",
    color_discrete_sequence=["#00A86B", "#FFA500"]
).show()
```

```python
# Minimum vs Maximum Stock
px.bar(
    vitamins,
    x="product_name",
    y=["min_stock_level", "max_stock_level"],
    barmode="group",
    title="Minimum vs Maximum Stock",
    color_discrete_sequence=["#6495ED", "#9370DB"]
).show()
```

```python
# Donut Chart
px.pie(vitamins,labels=vitamins["brand"],
    values=vitamins["product_id"], hole=0.5,
       title='Brand Distribution').show()
```

```python
# Scatter Plot - Price vs MRP
px.scatter(vitamins, x='price',
           y='mrp',
           color='brand'
          ,title='Price vs MRP').show()
```

```python
# Profit MArgin
px.bar(vitamins, x='brand',
       y='margin_percentage',color='brand',
       title='Profit Margin').show()
```

```python
# Price Distribution
px.histogram(vitamins,
             x='price', color ='price',
             title='Price Distribution').show()
```

```python
# Summary Dashboard
import plotly.express as px
from plotly.subplots import make_subplots
import plotly.graph_objects as go

fig = make_subplots(
    rows=3, cols=2,
    specs=[
        [{"type": "pie"}, {"type": "bar"}],
        [{"type": "bar"}, {"type": "histogram"}],
        [{"type": "scatter"}, {"type": "bar"}]
    ],
    subplot_titles=[
        "Brand Distribution",
        "Product Price",
        "Price vs MRP",
        "Price Distribution",
        "Price vs MRP Analysis",
        "Profit Margin"
    ]
)

# Pie chart
fig.add_trace(go.Pie(
    labels=vitamins["brand"],
    values=vitamins["price"],
    name="Brands"
), row=1, col=1)

# Product price
fig.add_trace(go.Bar(
    x=vitamins["product_name"],
    y=vitamins["price"],
    marker_color="green"
), row=1, col=2)

# Price vs MRP
fig.add_trace(go.Bar(
    x=vitamins["product_name"],
    y=vitamins["price"],
    name="Price",
    marker_color="blue"
), row=2, col=1)

fig.add_trace(go.Bar(
    x=vitamins["product_name"],
    y=vitamins["mrp"],
    name="MRP",
    marker_color="orange"
), row=2, col=1)

# Histogram
fig.add_trace(go.Histogram(
    x=vitamins["price"],
    marker_color="purple"
), row=2, col=2)

# Scatter plot
fig.add_trace(go.Scatter(
    x=vitamins["price"],
    y=vitamins["mrp"],
    mode="markers",
    marker_color="red"
), row=3, col=1)

# Profit margin
fig.add_trace(go.Bar(
    x=vitamins["product_name"],
    y=vitamins["margin_percentage"],
    marker_color="teal"
), row=3, col=2)

fig.update_layout(
    title="BLINKIT VITAMINS DASHBOARD",
    height=1000,
    showlegend=True
)

fig.show()
```
